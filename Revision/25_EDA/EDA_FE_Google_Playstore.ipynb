```python
# Convert Review Datatype to Int
df_copy['Reviews']=df_copy['Reviews'].astype(int)
```

```python
df_copy['Size'].unique()
```

```python
df_copy['Size'].isnull().sum()
```

```python
sum_count_values=df_copy['Size'].str.count(r'(\d+)M').sum()+df_copy['Size'].str.count(r'(\d+)k').sum()
```

```python
(df_copy['Size'] == 'Varies with device').sum() + sum_count_values
```

```python
df_copy['Size']=df_copy['Size'].str.replace('M','000').str.replace('k',
        '').replace('Varies with device',np.nan)

df_copy['Size']=df_copy['Size'].astype(float)
```

```python
df_copy.info()
```

```python
df['Installs'].unique()
```

```python
df['Price'].unique()
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Installs']=df_copy['Installs'].astype(int)
df_copy['Price']=df_copy['Price'].astype(float)
```

### Which Category has largest number of installations??

```python
df_copy.head()
```

```python
## Top 10 Categories having largest Installations
no_of_installs=df_copy.groupby('Category')['Installs'].sum().sort_values(ascending=False).reset_index()
no_of_installs['Installs']=no_of_installs['Installs']/1000000000 # converting into billions
top_10_installs=no_of_installs.head(10)

plt.figure(figsize = (14,10))
sns.set_context("talk")
sns.set_style("darkgrid")

ax = sns.barplot(x = 'Installs' , y = 'Category' , data = top_10_installs, palette='hls')
ax.set_xlabel('No. of Installations in Billions')
ax.set_ylabel('')
ax.set_title("Most Popular Categories in Play Store", size = 20)
```

### Insights
1. Out of all the categories "GAME" has the most number of Installations.
2. With 13 Billion+ Installations GAME is the most popular Category in Google App store

### What are the Top 5 most installed Apps in Each popular Categories ??

```python
dfa=df_copy.groupby(['Category','App'])['Installs'].sum().reset_index()
dfa=dfa.sort_values('Installs',ascending=False)
apps = ['GAME', 'COMMUNICATION', 'TOOLS', 'PRODUCTIVITY', 'SOCIAL' ]

sns.set_context("poster")
sns.set_style("darkgrid")

plt.figure(figsize=(40,30))

for i,app in enumerate(apps):
    df2 = dfa[dfa.Category == app]
    df3 = df2.head(5)
    plt.subplot(4,2,i+1)
    sns.barplot(data= df3,x= 'Installs' ,y='App', palette='Set2')
    plt.xlabel('Installation in Millions')
    plt.ylabel('')
    plt.title(app,size = 20)
    
plt.tight_layout()
plt.subplots_adjust(hspace= .3)
plt.show()
```

### Insights
- Most popular game is Subway Surfers.
- Most popular communication app is WhatsApp Messenger.
- Most popular tools app is Google.
- Most popular productivity app is Google Drive.
- Most popular social app is Instagram and Facebook.

### How many apps are there on Google Play Store which get 5 ratings??

```python
rating=df_copy.groupby(['Category','Installs','App'])['Rating'].sum().reset_index().sort_values('Installs',ascending=False)
top_rating=rating[rating['Rating']==5.0].reset_index().drop('index',axis=1)
print("Number of 5 rated apps are",top_rating.shape[0])
top_rating.head(5)
```

### Result
- There are 271 five rated apps on Google Play store
