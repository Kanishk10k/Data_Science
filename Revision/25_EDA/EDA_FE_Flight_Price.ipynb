```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df=pd.read_excel('Flight_price.xlsx')
df
```

```python
## Feature Engineering
df['Date']=df['Date_of_Journey'].str.split('/').str[0]
df['Month']=df['Date_of_Journey'].str.split('/').str[1]
df['Year']=df['Date_of_Journey'].str.split('/').str[2]
```

```python
df
```

```python
df.info()
```

```python
df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df['Year']=df['Year'].astype(int)
```

```python
df.info()
```

```python
## Drop Date of Journey

df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df
```

```python
df['Arrival_Time']=df['Arrival_Time'].str.split(' ').str[0]
```

```python
df['Arrival_hour']=df['Arrival_Time'].str.split(':').str[0]
df['Arrival_min']=df['Arrival_Time'].str.split(':').str[1]
```

```python
df.info()
```

```python
df['Arrival_hour']=df['Arrival_hour'].astype(int)
df['Arrival_min']=df['Arrival_min'].astype(int)
```

```python
df.info()
```

```python
df.drop('Arrival_Time',axis=1,inplace=True)
```

```python
df
```

```python
df['Dep_hour']=df['Dep_Time'].str.split(':').str[0]
df['Dep_min']=df['Dep_Time'].str.split(':').str[1]
```

```python
df.info()
```

```python
df['Dep_hour']=df['Dep_hour'].astype(int)
df['Dep_min']=df['Dep_min'].astype(int)
```

```python
df.info()
```

```python
df.drop('Dep_Time',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Total_Stops'].unique()
```

```python
df[df['Total_Stops'].isnull()]
```

```python
mode_stops=int(df['Total_Stops'].mode()[0][0])
```

```python
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0,'1 stop':1,'2 stops':2,
        '3 stops':3,'4 stops':4,np.nan:mode_stops})
```

```python
df['Total_Stops'].isnull().sum()
```

```python
df.info()
```

```python
df.drop('Route',axis=1,inplace=True)
```

```python
df['Duration']
```

```python
df['dur_hr'] = df['Duration'].str.extract(r'(\d+)h')
df['dur_min'] = df['Duration'].str.extract(r'(\d+)m')
```

```python
df.head()
```

```python
df['dur_hr'].unique()
```

```python
df['dur_hr'] = df['dur_hr'].fillna("0").astype(int)
df['dur_min'] = df['dur_min'].fillna("0").astype(int)
```

```python
df['total_duration_min'] = df['dur_hr'] * 60 + df['dur_min']
```

```python
df
```

```python
df.drop('Duration',axis=1,inplace=True)
```

```python
df.info()
```
