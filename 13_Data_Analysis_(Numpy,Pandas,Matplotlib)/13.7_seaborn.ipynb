```python
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
```

```python
tips.head()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(tips['total_bill'],tips['tip'],color='yellow',edgecolors='black')
plt.xlabel('Total Bill')
plt.ylabel('Tips')
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
```

```python
sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
```

```python
sns.lineplot(x='size',y='total_bill',data=tips)
plt.title('Lineplot of Size vs Total Bill')
plt.show()
```

```python
sns.barplot(x='day',y='total_bill',data=tips)
plt.title('Barplot of Day vs Total Bill')
plt.show()
```

```python
sns.boxplot(x='day',y='total_bill',data=tips)
plt.title('Boxplot of Day vs Total Bill')
plt.show()
```

```python
sns.violinplot(x='day',y='total_bill',data=tips)
plt.title('Violinplot of Day vs Total Bill')
plt.show()
```

```python
sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
sns.pairplot(tips)
```

```python
corr=tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True)
```
