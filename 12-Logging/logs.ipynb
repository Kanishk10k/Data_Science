```python
import logging

logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
```

```python
logging.debug("Debugging msg")
logging.info("info msg")
logging.warning("warning msg")
logging.error("Error msg")
logging.critical("critical msg")
```
