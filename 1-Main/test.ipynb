```python
import matplotlib.pyplot as pt
```

```python
x=[1,2,3,4,5]
x1=[i**2 for i in x]
x2=[(i+4)//2 for i in x]

pt.figure(figsize=(15,10))

pt.subplot(2,2,1)
pt.plot(x,x1,color='red')
pt.xlabel("Time")
pt.ylabel("Things")
pt.title("Plt-1")

pt.subplot(2,2,2)
pt.plot(x,x2,color='green')
pt.xlabel("Time")
pt.ylabel("Things")
pt.title("Plt-2")

pt.subplot(2,2,3)
pt.plot(x1,x2,color='blue')
pt.xlabel("Time")
pt.ylabel("Things")
pt.title("Plt-3")

pt.subplot(2,2,4)
pt.plot(x2,x1,color='purple')
pt.xlabel("Time")
pt.ylabel("Things")
pt.title("Plt-4")
```

```python
import sqlite3 as sq
```

```python
con=sq.Connection('eg.db')
con
cur=con.cursor()
```

```python
cur.execute('''
Create Table If Not Exists emp(
    id Integer Primary Key,
    name Text not null,
    age Integer,
    city Text
    )
''')
con.commit()
```

```python
data=[
    ('K',21,'U'),
    ('H',22,'R'),
    ('A',21,'U'),
    ('HT',20,'U'),
]

cur.executemany('''
    Insert Into emp(name,age,city)
        values(?,?,?)
''',data)

con.commit()
```

```python
cur.execute('Select * from emp')
row=cur.fetchall()
for i in row:
    print(i)
```

```python
con.close()
```
